```python

# DAY 3 PREPARATION
# Sweet Crumbs Home Bakery

business_name = "Sweet Crumbs Home Bakery"

catalogue = {
    "plum cake": {
        "price": 500,
        "category": "Cake",
        "eggless": True
    },
    "chocolate cake": {
        "price": 650,
        "category": "Cake",
        "eggless": True
    },
    "vanilla cake": {
        "price": 550,
        "category": "Cake",
        "eggless": True
    },
    "red velvet cake": {
        "price": 750,
        "category": "Cake",
        "eggless": True
    },
    "cupcake": {
        "price": 80,
        "category": "Dessert",
        "eggless": True
    }
}

print("Business:", business_name)
print("Catalogue:", catalogue)
```

```python

customer_questions = [
    "What cakes are available?",
    "What is the price of chocolate cake?",
    "Do you have eggless cakes?",
    "Can I order a cake for tomorrow?",
    "Do you provide home delivery?",
    "What is the price of six cupcakes?",
    "Can I customize a birthday cake?",
    "What flavours are available?",
    "How can I place an order?",
    "Can I cancel or modify my order?"
]

for i, question in enumerate(customer_questions, 1):
    print(f"{i}. {question}")
```

```python

def get_product(item: str) -> dict:
    """Get product details from the catalogue."""
    return catalogue.get(item.lower().strip(), {})


def get_price(item: str, quantity: int = 1) -> int:
    """Calculate the total product price."""
    product = get_product(item)

    if not product:
        return 0

    return product["price"] * quantity


print(get_product("plum cake"))
print("Total:", get_price("plum cake", 2))
```

```python

import json

SYSTEM_PROMPT = f"""
ROLE:
You are the assistant for {business_name}.

FACTS:
{json.dumps(catalogue)}

RULES:
- Never invent prices.
- Never provide false information.
- If unsure, ask the owner.

FORMAT:
- Answer in simple English.
- Be friendly and helpful.
- Keep answers short.
"""

print(SYSTEM_PROMPT)
```
